# Phase 1 — Create Datasets

### Objective

Create realistic datasets that represent a real business rather than working with one simple table.

For a sales analytics project, you could have datasets such as:

### 1. Customers Dataset

Contains information about customers.

Typical information:

* Customer ID
* Customer name
* Gender
* Age
* City
* State
* Region
* Registration date
* Customer segment

Example concept:

| Customer ID | Age | City  | Region | Registration Date |
| ----------- | --: | ----- | ------ | ----------------- |
| C001        |  24 | Patna | East   | 2024-01-12        |
| C002        |  31 | Delhi | North  | 2024-02-18        |

---

### 2. Products Dataset

Contains information about products.

Possible fields:

* Product ID
* Product name
* Category
* Sub-category
* Brand
* Cost price
* Selling price

---

### 3. Orders Dataset

This is usually the most important dataset.

Possible fields:

* Order ID
* Customer ID
* Product ID
* Order date
* Quantity
* Discount
* Payment method
* Sales channel
* Order status

---

### 4. Geography Dataset

Could contain:

* City
* State
* Region
* Population
* Market type

This isn't mandatory, but it gives you more opportunities for regional analysis.

---

### 5. Targets Dataset

This makes the project more realistic.

Possible fields:

* Month
* Region
* Sales target
* Profit target
* Order target

You can then compare **actual performance vs target**.

### Why create multiple datasets?

Because real-world analytics rarely gives you one perfectly organized table.

You'll need to understand:

> Which dataset contains what information, and how are these datasets related?

This introduces concepts such as **primary keys, foreign keys, relationships, and data integration**.

### Phase 1 Deliverable

By the end of Phase 1, you should have:

**4–5 clean-looking raw datasets with realistic relationships between them.**

Now you bring those datasets into your analysis environment.

* Missing values
* Duplicate records

## Step 2 — Understand every column

For every column, ask:

### What does this column represent?

For example:

**Customer ID**

→ Unique identifier for each customer.

**Quantity**

→ Number of units purchased.

**Discount**

→ Discount given on the order.

**Sales**

→ Revenue generated from the transaction.

You should never analyze a column without understanding its business meaning.

## Step 3 — Understand the structure

You should determine things like:

> Does every Order ID uniquely identify an order?

> Can one customer have multiple orders?

> Can one order contain multiple products?

> Does every Product ID exist in the Products dataset?

These questions become extremely important later.

## Step 4 — Initial exploratory analysis

Before cleaning anything, inspect the data as it currently exists.

You're trying to discover problems.

For example:

* Missing customer IDs
* Duplicate orders
* Incorrect dates
* Negative quantities
* Impossible ages
* Wrong region names
* Incorrect data types
* Extreme values

# Phase 3 — Data Cleaning + Validation

This is where the data becomes trustworthy.

---

# Part A — Data Cleaning

Your objective is:

> **Turn messy/raw data into analysis-ready data.**

---

## 1. Handle missing values

You need to determine why values are missing.

For example:

Customer's city is missing.

Possible approaches:

* Recover it from another dataset
* Replace it with a meaningful category
* Use statistical replacement where appropriate
* Remove the record if it's unusable

Don't blindly replace everything with zero.

---

## 2. Remove duplicates

You need to distinguish between:

### Genuine duplicate

Same transaction accidentally recorded twice.

### Legitimate repetition

A customer making two purchases.

These are completely different situations.

---

## 3. Fix data types

Examples:

Dates should behave like dates.

Quantity should behave like a number.

Price should behave like a numerical value.

Customer IDs should generally be treated as identifiers rather than mathematical quantities.

---

## 4. Standardize categorical values

You might discover:

* Delhi
* delhi
* DELHI
* New Delhi

These may represent the same location depending on your business rules.

Similarly:

* Electronics
* electronics
* ELECTRONICS

should be standardized.

---

## 5. Handle impossible values

Examples:

* Age = 250
* Quantity = -10
* Discount = 500%
* Selling price = negative
* Order date occurring before customer registration

These need investigation.

---

## 6. Handle outliers

Not every extreme value is an error.

For example:

One customer purchasing ₹500,000 worth of products might be:

* A genuine corporate customer
* A data entry mistake

You shouldn't automatically delete it.

You investigate first.

---

# Part B — Data Validation

Cleaning asks:

> **Can I fix this data?**

Validation asks:

> **Can I trust this data after fixing it?**

---

### Example validation rules

**Customer ID should be unique** in the customer master dataset.

**Every Order Customer ID should exist** in Customers.

**Every Order Product ID should exist** in Products.

**Quantity should be greater than zero.**

**Discount should fall within an acceptable range.**

**Order dates should be valid.**

**Selling price shouldn't be negative.**

---

### Think of validation as a quality-control system.

Your final dataset should satisfy business rules.

### Phase 3 Deliverable

A **clean + validated dataset**.

You should also maintain a record of:

* What problems were found
* What was changed
* Why it was changed
* How many records were affected

This is extremely valuable in a real analyst job.

# Part B — Feature Engineering

Feature engineering means:

> **Creating useful analytical variables from existing data.**

For example, you may derive:

### Revenue

How much money the transaction generated.

### Cost

How much the business spent.

### Profit

Revenue minus cost.

### Profit Margin

Profit relative to revenue.

### Order Month

Useful for monthly analysis.

### Order Year

Useful for yearly comparisons.

### Customer Age Group

For example:

* 18–25
* 26–35
* 36–50
* 50+

### Discount Amount

Actual monetary discount rather than only percentage.

### Average Order Value

Average amount spent per order.

---

### Why feature engineering matters

Raw data tells you:

> Customer purchased 4 units.

Engineered data can tell you:

> Customer generated ₹8,500 revenue and ₹2,100 profit.

That's much more useful for business analysis.

# Phase 8 — RFM Segmentation

This is one of the most important analytical sections of the project.

RFM stands for:

### R — Recency

**How recently did the customer purchase?**

A customer who purchased yesterday is generally more active than someone who purchased 10 months ago.

---

### F — Frequency

**How often does the customer purchase?**

Someone who purchased 20 times is more engaged than someone who purchased once.

---

### M — Monetary

**How much money has the customer spent?**

A customer spending ₹100,000 has a different value from someone spending ₹500.

---

# RFM Example

Imagine:

| Customer |  Recency | Frequency | Monetary |
| -------- | -------: | --------: | -------: |
| A        |   5 days |        18 |  ₹80,000 |
| B        |  20 days |         8 |  ₹30,000 |
| C        | 200 days |         2 |   ₹2,000 |

Customer A is clearly much more valuable.

---

# RFM Scoring

You assign scores to customers based on:

* Recency
* Frequency
* Monetary value

Usually, customers receive scores on a scale such as **1–5**.

Then you combine those scores.

---

# Possible Segments

You can create segments such as:

### Champions

Highly recent, frequent, high-spending customers.

### Loyal Customers

Purchase regularly.

### Big Spenders

Spend heavily but may not purchase frequently.

### New Customers

Recently acquired.

### At Risk

Previously valuable but haven't purchased recently.

### Lost Customers

Long inactive.

---

# Why RFM matters

Instead of saying:

> "We have 5,000 customers."

You can say:

> "We have 420 high-value customers, 850 loyal customers, 600 new customers, and 300 customers at risk of churn."

That's significantly more useful to a business.

# Phase 10 — Automatic Insights

This is where the project becomes much more interesting.

Instead of manually looking at every chart and writing observations, you build a system that identifies important patterns.

---

## Example

Your analysis discovers:

**Revenue increased 24%**

The system could produce an insight like:

> Revenue increased significantly compared with the previous period.

---

Another example:

Suppose:

**Revenue increased**

but

**Profit decreased**

The system should recognize that something is wrong.

Possible reasons:

* Higher discounts
* Higher costs
* Product mix changes
* Lower-margin products selling more

---

## Possible automatic insight rules

You can detect:

### Revenue growth

Revenue increased significantly.

### Revenue decline

Revenue dropped significantly.

### Profit decline

Profit is decreasing despite sales growth.

### High-performing product

A product contributes unusually high revenue or profit.

### Low-performing region

A region is underperforming against the company average or target.

### High discount

Discount levels are unusually high.

### Customer concentration

A small group of customers generates a large percentage of revenue.

### Churn risk

A significant group of valuable customers hasn't purchased recently.

### Target achievement

A region/product has exceeded or missed its target.

---

# Example of an Insight Engine

The analytical logic could conceptually work like:

**Metric**

↓

**Compare against benchmark**

↓

**Detect significant difference**

↓

**Classify pattern**

↓

**Generate human-readable insight**

For example:

**Profit margin**

↓

Current margin = 9%

↓

Previous margin = 15%

↓

Decrease = 6 percentage points

↓

**Insight:**

> Profit margin has declined considerably compared with the previous period.

---

# Important distinction

Automatic insights shouldn't just repeat numbers.

Bad:

> "Revenue was ₹50 lakh."

Better:

> "Revenue increased 18% compared with the previous period."

Best:

> "Revenue increased 18%, but profit grew only 3%, suggesting margin pressure that may be linked to higher discounts or costs."

The third one actually helps decision-making.

An **automatic business-insight engine** that converts analytical results into readable observations.
